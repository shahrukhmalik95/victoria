Buster's Sports Bar & Grill runs a busy draft beer program, pouring a house lager through a long-draw, CO2-driven tap system. ### Q3: Briefly describe why a CUSUM and EWMA chart would be appropriate in this scenario, given the general manager's specific concern about the aging CO2 regulator. What are the limitations of these charts?

### ANSWER
CUSUM and EWMA charts are appropriate because they help detect small, gradual changes in beer pour volume caused by the aging CO₂ regulator. Overall, these charts help the bar catch pouring issues early, reducing waste and maintaining customer satisfaction.

### Q4: Read the Phase I data into R. Calculate and report the subgroup means and ranges, and use them to estimate the process mean and process standard deviation.

### ANSWER

```r
## Q4 Code ##


install.packages("qcc", repos = "http://cran.us.r-project.org")

library(tidyverse)
library(readxl)
library(qcc)

phase1 <- read_excel("/content/BustersSportsBar_PhaseI.xlsx")
##phase1
```

```r

pours <- as.matrix(phase1[, 2:6])

#  subgroup mean for each shift
phase1$Mean <- rowMeans(pours)

#  subgroup range (Maximum - Minimum)
phase1$Range <- apply(pours, 1, function(x) max(x) - min(x))

# subgroup means and ranges
print(phase1[, c("Shift", "Mean", "Range")])

# overall process mean
X_double_bar <- mean(phase1$Mean)

# Calculate average range
R_bar <- mean(phase1$Range)

# Estimate process standard deviation
# d2 = 2.326 for subgroup size n = 5
sigma_hat <- R_bar / 2.326

# Display final results
cat("Process Mean:", round(X_double_bar, 4), "oz\n")
cat("Average Range:", round(R_bar, 4), "oz\n")
cat("Process Standard Deviation:", round(sigma_hat, 4), "oz\n")
```

Based on the Phase I data collected over 25 shifts, the estimated process mean is 16.0386 oz, the average subgroup range is 0.1795 oz, and the estimated process standard deviation is 0.0772 oz. The average pour volume is slightly higher than the target of 16.00 oz, indicating a small amount of overpouring on average. These values provide initial estimates for evaluating the pouring process and developing CUSUM and EWMA control charts.
